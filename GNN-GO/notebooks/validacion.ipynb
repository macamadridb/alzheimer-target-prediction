```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
# Configuración general de gráficos
sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
```

```python
base_output_dir = os.path.join("..", "notebooks","GNN-GO", "output")
modulos_df = os.path.join(base_output_dir, "resultados_modulos.csv")
go_df = os.path.join(base_output_dir, "proteinas_con_go_representativo.csv")

base_input_dir = os.path.join("..", "notebooks","GNN-GO", "input")
metadata_path = os.path.join(base_input_dir, "metadata_proteina.csv")


#metadata_path = "../data/Input/metadata_proteins.csv"
```

```python
# --- 2. Cantidad de proteínas por módulo ---
modulos_df['num_proteins'].plot(kind='bar', color='skyblue')
plt.title("Número de proteínas por módulo")
plt.xlabel("Módulo")
plt.ylabel("Número de proteínas")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# --- 3. Distribución de proteínas DEG por módulo ---
# Convertir string de dict a diccionario real
import ast
modulos_df['deg_distribution'] = modulos_df['deg_distribution'].apply(ast.literal_eval)

deg_expanded = modulos_df[['module_id', 'deg_distribution']].explode('deg_distribution')
deg_rows = []
for idx, row in modulos_df.iterrows():
    for status, perc in row['deg_distribution'].items():
        deg_rows.append({'module_id': row['module_id'], 'DEG': status, 'Porcentaje': float(perc.strip('%'))})

deg_df = pd.DataFrame(deg_rows)

sns.barplot(data=deg_df, x='module_id', y='Porcentaje', hue='DEG')
plt.title("Distribución de proteínas DEG por módulo")
plt.xlabel("Módulo")
plt.ylabel("% de proteínas")
plt.xticks(rotation=90)
plt.legend(title="Estado DEG", bbox_to_anchor=(1.05, 1))
plt.tight_layout()
plt.show()
```

```python
# --- 4. Distribución de Target Group por módulo ---
modulos_df['target_group_distribution'] = modulos_df['target_group_distribution'].apply(ast.literal_eval)

tg_rows = []
for idx, row in modulos_df.iterrows():
    for tg, perc in row['target_group_distribution'].items():
        tg_rows.append({'module_id': row['module_id'], 'Target_Group': tg, 'Porcentaje': float(perc.strip('%'))})

tg_df = pd.DataFrame(tg_rows)

sns.barplot(data=tg_df, x='module_id', y='Porcentaje', hue='Target_Group')
plt.title("Distribución de Target Groups por módulo")
plt.xlabel("Módulo")
plt.ylabel("% de proteínas")
plt.xticks(rotation=90)
plt.legend(title="Target Group", bbox_to_anchor=(1.05, 1))
plt.tight_layout()
plt.show()
```

```python
# --- 5. GO representativos más frecuentes ---
top_go_terms = go_df['GO_term'].value_counts().head(15)

top_go_terms.plot(kind='barh', color='salmon')
plt.title("Términos GO representativos más frecuentes")
plt.xlabel("Número de proteínas")
plt.ylabel("GO_term")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
# matriz confusión
# Unir datos para obtener Target Group y DEG por proteína
merged_df = go_df.merge(metadata_df, on='proteina')

# Crear tabla de contingencia
conf_matrix = pd.crosstab(merged_df['Target_group'], merged_df['DEG'])

# Visualizar con heatmap
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues')
plt.title("Matriz de Confusión: Target Group vs DEG")
plt.xlabel("Estado DEG")
plt.ylabel("Target Group")
plt.tight_layout()
plt.show()
```
